## Q3A 传感器评分器的 NumPy 前向、反向与灵敏度图

**30分钟，100分；依据第3章与Homework02 Exercise 7缩小规模新编。不是原作业30分题的完整搬用。**

一个双输入传感器的标量评分器为 $f(x)=v^T\tanh(Wx+b)+c$。
固定 `W=[[.5,-1],[1,.25]]`、`b=[.1,-.2]`、`v=[1.5,-.75]`、`c=.3`。
从导入库开始写一个完整程序，不使用 PyTorch，不做代码补全。

1. **20分/约5分钟**：创建参数及四行输入 `X=[[0,0],[1,0],[0,1],[-1,.5]]`，打印网络层形状和9个参数；定义 `score3a(x)` 返回字典 `f,dx,dW,db,dv,dc`。
2. **35分/约10分钟**：手动前向及反传，四个输入均打印函数值与全部梯度，权重梯度必须是外积；保存四个字典的列表 `reports3a`。
3. **25分/约7分钟**：在 `x=[1,0]` 对 `x1` 和 `W[0,0]` 用 $h=10^{-6}$ 中心差分校准；输出解析值、数值及绝对误差，保存 `fd_errors3a`（长度2）。扰动参数后必须恢复。
4. **20分/约8分钟**：令第二输入固定为0，第一输入在 `[-2,2]` 取101点，保存 `output/Q3A.png`：画 `f((t,0))` 与 `df/dx1` 两条曲线，英文轴名、图例；打印最大绝对灵敏度。

评分：数值正确并不替代手动反传实现要求；只硬编码四点结果不得反传分。测试与教学独立，不依赖前序单元。


```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
W3a=np.array([[.5,-1.],[1.,.25]],dtype=float)
b3a=np.array([.1,-.2]);v3a=np.array([1.5,-.75]);c3a=.3
X3a=np.array([[0.,0.],[1.,0.],[0.,1.],[-1.,.5]])
def score3a(x):
    x=np.asarray(x,dtype=float)
    h=np.tanh(W3a@x+b3a)
    p=v3a*(1-h*h)
    return dict(f=float(v3a@h+c3a),dx=W3a.T@p,dW=np.outer(p,x),
                db=p,dv=h,dc=1.)
print('W,b,v,c shapes:',W3a.shape,b3a.shape,v3a.shape,(),'parameters',W3a.size+b3a.size+v3a.size+1)
reports3a=[score3a(x) for x in X3a]
for x,r in zip(X3a,reports3a): print('input',x,r)
hfd3a=1e-6;x3a=X3a[1].copy()
xp=x3a.copy();xm=x3a.copy();xp[0]+=hfd3a;xm[0]-=hfd3a
fdx=(score3a(xp)['f']-score3a(xm)['f'])/(2*hfd3a)
saved=W3a[0,0]
W3a[0,0]=saved+hfd3a;fp=score3a(x3a)['f']
W3a[0,0]=saved-hfd3a;fm=score3a(x3a)['f']
W3a[0,0]=saved
fdw=(fp-fm)/(2*hfd3a)
fd_errors3a=np.abs(np.array([fdx,fdw])-np.array([reports3a[1]['dx'][0],reports3a[1]['dW'][0,0]]))
print('x1 analytic/numeric/error',reports3a[1]['dx'][0],fdx,fd_errors3a[0])
print('W00 analytic/numeric/error',reports3a[1]['dW'][0,0],fdw,fd_errors3a[1])
grid3a=np.linspace(-2,2,101)
vals3a=[score3a(np.array([t,0.])) for t in grid3a]
plt.figure();plt.plot(grid3a,[r['f'] for r in vals3a],label='score')
plt.plot(grid3a,[r['dx'][0] for r in vals3a],label='input sensitivity')
plt.xlabel('first input');plt.ylabel('value');plt.legend()
Path('output').mkdir(exist_ok=True);plt.savefig('output/Q3A.png');plt.close()
print('maximum absolute sensitivity',max(abs(r['dx'][0]) for r in vals3a))
```

### 解析与点评

**评分点评**：20分检查参数方向/总量9；35分检查 `p=v*(1-h²)`、`dW=p xᵀ`、`dx=Wᵀp` 及其余梯度；25分检查两个非零中心差分与参数恢复；20分检查101点两曲线、英文标签和PNG。`(1,0)` 的第二列权重梯度为0，是输入分量为0的边界；公开自检另用非零第二输入检查这一列，避免只会处理零值掩盖实现错误。


### 自检（作答后运行）

这些是公开的代表性检查，不替代评分细则、图表检查与书面解释；通过不代表已满足全部要求。


```python
import numpy as np
from pathlib import Path
assert len(reports3a)==4 and np.max(fd_errors3a)<1e-7
Wref=np.array([[.5,-1.],[1.,.25]]);bref=np.array([.1,-.2]);vref=np.array([1.5,-.75])
for x in [np.array([.3,-.4]),np.array([-1.2,2.])]:
    r=score3a(x);hh=np.tanh(Wref@x+bref);pp=vref*(1-hh*hh)
    assert abs(r['f']-(vref@hh+.3))<1e-12
    assert np.allclose(r['dx'],Wref.T@pp) and np.allclose(r['dW'],np.outer(pp,x))
    assert np.allclose(r['db'],pp) and np.allclose(r['dv'],hh) and r['dc']==1
assert Path('output/Q3A.png').stat().st_size>1000

print("自检通过")
```
